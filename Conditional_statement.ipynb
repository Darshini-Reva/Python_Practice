### if -elif -else condition

```python
## syntax       # if-elif-else ladder

if condition1:  
    # Code for condition1 if True 
elif condition2:
    # Code for condition2 if True (only checked if condition1 is False)
elif condtion3:
    # Code for condition3 if True (only checked if condition1 and condition2 is False)
elif:
    # condition
elif:
    # condition
elif:
    # condition
elif:
    # condition
elif:
    # condition
else:
    # code if all condition are false(optional)
```

```python
## Grading system Grade "A" > 90, grade "B" >= 80 ,grade "C" >= 70  , "D" >= 60, else Fail

score = 75

if score >= 90:      # False
    grade = "A"
    print("Excellent work")
elif score >= 80:        #False
    grade = "B"
    print("Good job!")
elif score >= 70:    # True
    grade = "C"
    print("Good effort")
elif score >= 60:
    grade = "D"
    print("You passed but you need to improve")
else: 
    grade = "F"
    print("Failed. please retake the course")
    
print(f"Your score is {score}, Grade {grade}")
```
