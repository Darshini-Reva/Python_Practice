### Creating, accessing ([] , get()) , Adding / updating

```python
learning_paths = {
    "Data Science": ["Python", "Pandas", "ML"],
    "Data Engineering": ["ADF", "ADLS", "Databricks"],
}
student_ratings = {"Alice": 4.5, "Bob": 4.8}
```

```python
print(learning_paths)
```

```python
### accessing the tools for data science using []

ds_tools = learning_paths ["Data Science"]
print(ds_tools)
```

```python
### accessing the tools for non-existing using[]

cloud_tools = learning_paths["Cloud computing"]  ## it will give key error
cloud_tools
```

```python
### accessing the tools for data science using get()

de_tools = learning_paths.get("Data Engineering")
de_tools
```

```python
### accessing the tools for non-existing using get() with default value

ml_tools = learning_paths.get("Machine Learning",[])  ## it will give alternative value
ml_tools
```

```python
ml_tools = learning_paths.get("Machine Learning",["Linear Regression", "Decision tree"])  ## it will give alternative value
ml_tools
```

```python
cloud_tools = learning_paths.get("Cloud computing" , 100)  ## it will give key error
cloud_tools
```
