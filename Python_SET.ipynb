#### Basic set operators

```python
## add

fruits = {"apple", "banana"}
fruits.add("orange")
fruits
```

```python
fruits = {"apple", "banana"}
fruits.update(["mango", "grapes", "kiwi"])
fruits.update(["pear"])   # list
fruits.update(("plum"))   # tuple
print(fruits)
```

```python
colors.discard("purple")   # it will not throw error
print(colors) 
```

### Mathematical set opertions

```python
## Union

set_a = {1,2,3,4}
set_b = {3,4,5,6}
result = set_a| set_b
print(result)
```

```python
set_a = {1,2,3,4}
set_b = {3,4,5,6}
result = set_a.union (set_b)
print(result)
```

```python
### union with multiple set

set_a = {1,2,3,4}
set_b = {3,4,5,6}
set_c = {6,7,8,9}
result_mutliple = set_a | set_b | set_c
print(result_mutliple)
```

```python
abishek_hobbies = {"reading", "gaming", "cooking"}
manasa_hobbies = {"gaming", "painting", "dancing"}

all_hobbies = abishek_hobbies | manasa_hobbies

print(all_hobbies)
```

```python
## Intersection

set_a = {1,2,3,4}
set_b = {3,4,5,6}

set_a & set_b
```

```python
## using intersection() method

set_a.intersection(set_b)
```

```python
## Difference 

set_a = {1,2,3,4}
set_b = {3,4,5,6}

results = set_a - set_b
print(results)
```

```python
results = set_b - set_a
print(results)
```

```python
## using difference() method

set_a = {1,2,3,4}
set_b = {3,4,5,6}

results = set_a.difference(set_b)
print(results)
```

```python
### Symmetric difference

set_a = {1,2,3,4}
set_b = {3,4,5,6}
```

```python
result = set_a ^ set_b
print(result)
```

```python
set_a = {1,2,3,4}
set_b = {3,4,5,6}

results1 = set_a.symmetric_difference(set_b)
print(results1)
```

### Set comparison operators

```python
## subset 

set_a = {1,2,3}
set_b = {1,2,3,4,5}
print(set_a.issubset(set_b))
```

```python
print(set_a <= set_b)
```

```python
## disjoint

set_a = {1,2,3}
set_b = {4,5,6}
set_c = {3,4,5}

print(set_a.isdisjoint(set_b))
print(set_a.isdisjoint(set_c))
```
