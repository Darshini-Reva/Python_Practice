```python
### common operations and usage:

# - literals and comparisons

is_data_science_active = True
is_azure_devops_course_full = False
is_python_essential = True

total_students = 1500
capacity = 1500
average_rating = 4.8
min_rating = 4.5
```

```python
### equality check (==)
# = - for assignment
# == - for comparison
# is - for identity

at_capacity = (total_students == capacity)
at_capacity
```

```python
# inequality check (!=)

has_vacanies = (total_students != capacity)
has_vacanies
```

```python
# greater than check(>)

rating_is_good = (average_rating > 4.0)
rating_is_good
```

```python
rating_is_good = (average_rating > min_rating)
rating_is_good
```

```python
# Less than check (<)

need_improvement = (average_rating < 4.0)
print (need_improvement)
type(need_improvement)
```
