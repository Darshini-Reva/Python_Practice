#### list.sort(key = None, reverse = False)

```python
course_list = ["Data Engineering", "Azure DevOps", "Data Science", "Python Basics"]
student_counts = [350, 420, 380, 500, 450]
mixed_case_courses = ["azure DevOps", "Python basics", "Data science"]
```

```python
##sort course list alphabetically

course_list = ["Data Engineering", "Azure DevOps", "Data Science", "Python Basics"]
print(course_list)
course_list.sort()
print(course_list)
```

```python
course_list = ["Data Engineering", "Azure DevOps", "Data Science", "Python Basics"]
print(f" course list before sorting: {course_list}")
course_list.sort()
print(f" course list after sorting: {course_list}")
```

```python
course_list  # modified original synatx
```

```python
student_counts = [350, 420, 380, 500, 450]  ## ascending

print(f" students list before sorting: {student_counts}")
student_counts.sort()
print(f" students list after sorting: {student_counts}")
```

```python
student_counts = [350, 420, 380, 500, 450]  ## descending

print(f" students list before sorting: {student_counts}")
student_counts.sort(reverse=True)
print(f" students list after sorting: {student_counts}")
```

```python
mixed_case_courses = ["azure DevOps", "ball" ,"Python basics", "Data science"]

mixed_case_courses.sort()
print(mixed_case_courses)
```

```python
mixed_case_courses = ["azure DevOps", "ball" ,"Python basics", "Data science"]

mixed_case_courses.sort(key= str.lower)
print(mixed_case_courses)
```

```python
courses = ["SQL", "Python", "Azure DevOps", "Git"]
print(courses)
courses.sort(key = len)
print(courses)
```

```python
courses = ["SQL", "Python", "Azure DevOps", "Git"]
print(courses)
courses.sort(key = len, reverse=True)
print(courses)
```

```python
batches = [[102, 'DE'], [101, 'DS'], [103, 'ADO']]
print(batches)
batches.sort()
print(batches)
```

```python
batches = [[102, 'DE'], [101, 'DS'], [103, 'ADO']]
print(batches)
batches.sort(key=lambda x:x[1])   ## to take second element in list
print(batches)
```

```python
batches = [[102, 'DE', "C"], [101, 'DS', "B"], [103, 'ADO', "A"]]
print(batches)
batches.sort(key=lambda x:x[2])   ## to take third element in list
print(batches)
```
